# KO / PEP Engle-Granger slice

**Hypothesis (this notebook, this slice):**

1. KO and PEP daily adjusted closes are cointegrated, so the OLS residual (spread) is mean-reverting.
2. A z-score of that spread, using a hedge ratio from Engle-Granger, can be traded after explicit spread + slippage costs.
3. A date-based walk-forward split will show whether out-of-sample Sharpe is real or only in-sample fit.

Trading uses **Engle-Granger only**. Johansen is printed as a diagnostic; we do not invent VECM results and we do not trade its eigenvectors.

Defaults (`z_enter=2`, `z_exit=0.5`, `5+5` bps, 70% train) are **not optimized**. Nothing here is live PnL or alpha.

If yfinance fails, we load `data/sample_ko_pep.csv`. That file is a **synthetic sample**, not KO/PEP market data. `python data/make_sample.py` regenerates it.

## How to read this

- If **test_costed Sharpe** is near zero or negative while **train_uncosted** looks fine, the in-sample fit was the whole story.
- Costs strictly reduce Sharpe on the same path by construction of the backtest; the interesting question is whether costed OOS Sharpe is still positive.
- A single pair, one split, Yahoo (or synthetic) prices, and unoptimized thresholds are not a strategy. They are a check that the residual machinery is wired correctly.

Regime breaks, borrow, next-open execution, and a real cost tape are out of scope. Yahoo is not research-grade; survivorship and liquidity are not modelled.
